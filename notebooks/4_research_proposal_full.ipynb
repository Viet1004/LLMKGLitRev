```python
# Run the multi-agent supervisor agent
from utils import format_messages
from langchain_core.messages import HumanMessage
from llmkglitrev.samples import sample_1, sample_2
from llmkglitrev.agents.research_proposal_generator import proposal_generator_agent
from llmkglitrev.agents.states import AgentState, SupervisorState
from langgraph.types import Command
from rich.console import Console
from rich.panel import Panel

console = Console()

# Configure with thread_id for interrupt support
config = {"configurable": {"thread_id": "test-notebook-1"}}

# Invoke the agent with proper config parameter
result = await proposal_generator_agent.ainvoke(
    {
        "messages": [HumanMessage(content=f"{sample_2}.")]
    },
    config=config
)

# Check if interrupted
if "__interrupt__" in result:
    print("⚠️ Agent was interrupted - waiting for human feedback")
    print("\nInterrupt data:")
    interrupt_data = result["__interrupt__"][0].value
    print(f"Question: {interrupt_data['question']}")
    print(f"\nProposals: {interrupt_data['proposals']}")
    
    # Collect feedback
    feedback = input("\nYour feedback (or press Enter to approve all): ").strip()
    if not feedback:
        feedback = "All proposals look good. Please proceed."
    
    # Resume with feedback
    result = await proposal_generator_agent.ainvoke(
        Command(resume=feedback),
        config=config
    )

# Now final_proposal should exist
if "final_proposal" in result:
    console.print(Panel(result['final_proposal'], title="📝 Final Research Proposal", border_style="cyan"))
else:
    print("❌ Error: final_proposal not found in result")
    print(f"Available keys: {list(result.keys())}")
```

## Option 2: Simpler Approach with Helper Function

A more reliable approach for notebooks that avoids input issues:


```python
# Option 2: Helper function for cleaner code
from llmkglitrev.samples import sample_1, sample_2
from llmkglitrev.agents.research_proposal_generator import proposal_generator_agent
from langchain_core.messages import HumanMessage
from langgraph.types import Command
from rich.console import Console
from rich.panel import Panel

console = Console()

async def run_research_with_feedback(query, thread_id="test-notebook-2"):
    """Helper function to run research with interrupt handling"""
    config = {"configurable": {"thread_id": thread_id}}
    
    # Initial invocation
    result = await proposal_generator_agent.ainvoke(
        {"messages": [HumanMessage(content=query)]},
        config=config
    )
    
    # Handle interrupt if present
    if "__interrupt__" in result:
        console.print("\n[bold yellow]⚠️  Agent interrupted - requesting feedback[/bold yellow]\n")
        interrupt_data = result["__interrupt__"][0].value
        
        console.print(Panel(interrupt_data['question'], title="❓ Question", border_style="yellow"))
        
        console.print("\n[bold cyan]📋 Proposed Research Directions:[/bold cyan]")
        for i, proposal in enumerate(interrupt_data['proposals'], 1):
            console.print(f"   {i}. {proposal}")
        
        # Collect feedback (or auto-approve)
        feedback = "All proposals look excellent. Please proceed with all of them."
        console.print(f"\n[bold green]✅ Feedback:[/bold green] {feedback}\n")
        
        # Resume
        result = await proposal_generator_agent.ainvoke(
            Command(resume=feedback),
            config=config
        )
    
    return result

# Run the research
result = await run_research_with_feedback(f"{sample_2}.")

# Display final proposal
if "final_proposal" in result:
    console.print(Panel(result['final_proposal'], title="📝 Final Research Proposal", border_style="cyan"))
```
